```python
# Task C1 - Translation (Fixed for new library version)
!pip install transformers sentencepiece -q

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

# 1. Load the model and tokenizer directly
print("Loading model... (1-2 mins)")
model_name = "Helsinki-NLP/opus-mt-en-ur"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# 2. Sentences to translate
sentences = [
    "Hello, how are you?",
    "I am a student.",
    "Machine learning is interesting.",
    "The weather is nice today.",
    "Submit the assignment on time."
]

print("\n--- TRANSLATION EXAMPLES ---")

# 3. Translate each sentence
for i, text in enumerate(sentences):
    # Convert the text into numbers the model understands
    inputs = tokenizer(text, return_tensors="pt")

    # Generate the Urdu translation
    translated_tokens = model.generate(**inputs, max_length=100)

    # Convert the numbers back into readable text
    urdu_text = tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)[0]

    print(f"Example {i+1}:")
    print(f"English: {text}")
    print(f"Urdu: {urdu_text}")
    print("-" * 40)
```
